```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
```

```python
X,y = load_diabetes(return_X_y = True)
```

*10 input cols, 1 output col*

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2,random_state=4
)
```

solver :
1. sparse_cg,sag,saga -> uses gradient descent
2. svd,cholesky,lsqr -> uses ordinary least square (formula based)

```python
from sklearn.linear_model import Ridge
reg = Ridge(alpha=0.1,solver='cholesky')
```

```python
reg.fit(X_train,y_train)
```

```python
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
class MyndimRidge:
  # n -> the no. of cols
  # m -> the no of rows

  def __init__(self,alpha=0.1):

    self.alpha = alpha
    self.coef_ = None
    self.intercept_ = None

  def fit(self,X_train,y_train):
    X_train = np.insert(
        X_train,0,1,axis=1
    )
    # shape of identity = shape of XtransposeX
    # i.e ((n+1) * m * m * (n+1)) = ((n+1),(n+1))
    # identity is a square matrix
    I = np.identity(X_train.shape[1])
    I[0][0] = 0
    W = (np.linalg.inv(
        (np.dot(X_train.T,X_train)) + (self.alpha * I)
    )).dot(X_train.T).dot(y_train)
    # now W will contain (n+1,1)
    # W0,W1,W2,...Wn
    self.intercept_ = W[0]
    self.coef_ = W[1:]


  def predict(self,X_test):
    return np.dot(X_test,self.coef_) + self.intercept_
```

*why I[0][0] = 0 ?*
 *because overfitting happens due to training data and not based on offset values, so we don't apply regularisation on intercept*

```python
reg = MyndimRidge()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
print(r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```
