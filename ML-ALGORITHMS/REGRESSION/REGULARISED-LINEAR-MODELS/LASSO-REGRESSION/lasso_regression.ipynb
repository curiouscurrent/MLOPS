```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
X,y = make_regression(
    n_samples=100,n_features=1,n_informative=1,n_targets=1,noise=20,random_state=13
)
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2
)

plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X_train,y_train)
print(reg.coef_)
print(reg.intercept_)
```

**In lasso : if you make the alpha very high, coefficients will become 0**

**For intermediate values of alpha -> lasso will do feature selection by making the coefficient of less important columns as 0, (i.e the column's that are not important to calculate y)**

**For very high values of alpha -> all the coefficients become 0 -> underfitting**

*alpha is inversely proportional to coefficient values*

```python
alphas = [0,1,5,10,30]
plt.figure(figsize=(12,6))
plt.scatter(X,y)
for i in alphas:
  L = Lasso(alpha=i)
  L.fit(X_train,y_train)
  plt.plot(X_test,L.predict(X_test),label='alpha={}'.format(i))
plt.legend()
plt.show()
```

```python
# non linear data
m=100
x1 = 5 * np.random.rand(m,1) - 2
x2 = 0.7 * x1 ** 2 - 2*x1 + 3 + np.random.randn(m,1)

plt.scatter(x1,x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Lasso
```

```python
def get_preds_lasso(x1,x2,alpha):
  model = Pipeline([
      ('poly_features',PolynomialFeatures(degree=16)),
      ('lasso',Lasso(alpha=alpha))
  ])
  # After applying poly_features on input cols
  # we get x1^0, x1^1, x1^2,....x1^16 -> 17 input cols
  # since its a high dim data, there is chance of overfitting
  # apply lasso to shrink the coef value
  model.fit(x1,x2)
  # now we get the values of the coeffs of 17 input cols
  return model.predict(x1)

```

alpha = 0 -> case of linear regression -> overfitting possible

alpha = 1 -> underfitting

alpha = 0.1 -> optimum -> intermediate

```python
alphas = [0,0.1,1]
cs = ['r','g','b']

plt.figure(figsize=(10,6))
plt.plot(x1,x2,'b+',label='Datapoints')

for alpha,c in zip(alphas,cs):
  # first iter -> 0,'r'
  preds = get_preds_lasso(x1,x2,alpha)
  # plot
  plt.plot(
      sorted(x1[:,0]), preds[np.argsort(x1[:,0])], c,
      label = 'Alpha : {}'.format(alpha)
  )
plt.legend()
plt.show()
```

red - overfitting

green - optimum

blue - underfitting
