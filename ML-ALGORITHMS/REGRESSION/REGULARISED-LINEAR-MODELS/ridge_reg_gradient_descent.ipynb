Ridge regression -> L2 regularisation

```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2,random_state=4
)
```

SGDRegressor implements gradient descent

```python
from sklearn.linear_model import SGDRegressor
reg = SGDRegressor(
    penalty='l2', # l2 -> ridge regularisation
    max_iter=500,
    eta0=0.1,
    learning_rate='constant', # not using learning schedule
    alpha = 0.001 # regularisation hyperparamter

)
```

```python
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
print("R2 score : ",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```

```python
class MyRidgeGD:
  def __init__(self,epochs,learning_rate,alpha):
    self.epochs = epochs
    self.learning_rate = learning_rate
    self.alpha = alpha
    self.coef_ = None
    self.intercept_ = None

  def fit(self,X_train,y_train):
    # Initial weights
    # initially the coefficients W1..Wn -> set as 1
    # set the intercept W0 as 0
    self.intercept_ = 0
    self.coef_ = np.ones(X_train.shape[1])
    X_train = np.insert(X_train,0,1,axis=1)
    # weight matrix -> thetha
    # first value will be the bias W0
    thetha = np.insert(self.coef_,0,self.intercept_)

    # calculate gradients
    # i.e partial derivatives of L w.r.t W0,W1,..Wn
    for i in range(self.epochs):
      thetha_der = np.dot(X_train.T,X_train).dot(thetha) - np.dot(X_train.T,y_train) + (self.alpha * thetha)
         # update weights using gradient descent
      thetha = thetha - self.learning_rate * thetha_der

    self.coef_ = thetha[1:]
    self.intercept_ = thetha[0]



  def predict(self,X_test):
    return np.dot(X_test,self.coef_) + self.intercept_
```

```python
reg = MyRidgeGD(
    epochs=500,
    alpha=0.01,
    learning_rate=0.005
)
```

```python
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
print("R2 Score : ",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```
