When we have more than 1 input column, and only one output column -> we use multiple linear regression.

Simple Linear regression is a specialisation of multiple LR, where there is only one input column

In sklearn -> we have 2 methods for multiple LR:
1. OLS (closed form solution -> via formula -> LinearRegression class)

2. Gradient descent (non-closed form solution -> via SGDRegressor class)

```python
from sklearn.datasets import make_regression
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

*2 input columns, 1 output column*

```python
X,y = make_regression(
    n_samples = 100,
    n_features = 2,
    n_informative = 2,
    n_targets = 1,
    noise = 50
)
```

```python
X[0]
```

```python
df = pd.DataFrame(
    {
        'feature1' : X[:,0],
        'feature2' : X[:,1],
        'target' : y
    }
)
```

```python
df.shape
```

```python
df.head()
```

```python
# scatter the datapoints
fig = px.scatter_3d(
    df,
    x = 'feature1',
    y= 'feature2',
    z = 'target'
)
fig.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2, random_state=3
)
```

Upon fitting the model, i.e training, we find the values of coefficients and intercept

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
print("MAE", mean_absolute_error(y_test,y_pred))
print("MSE", mean_squared_error(y_test,y_pred))
print("R2 Score",r2_score(y_test,y_pred))
```

```python
x = np.linspace(-5,5,10)
y = np.linspace(-5,5,10)
xGrid, yGrid = np.meshgrid(y,x)
final = np.vstack(
    (
        xGrid.ravel().reshape(1,100),
        yGrid.ravel().reshape(1,100)
    )
).T
z_final = lr.predict(final).reshape(10,10)
z = z_final
```

```python
fig = px.scatter_3d(
    df, x ='feature1', y='feature2', z='target'
)
fig.add_trace(
    go.Surface(
        x = x, y=y, z=z
    )
)
fig.show()
```
